```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

p=list(Path(".").rglob("iris.csv"))
if not p: raise FileNotFoundError("Extract the practical datasets ZIP beside this notebook.")
df=pd.read_csv(p[0]); X=df.drop(columns="target"); y=df["target"]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
sc=StandardScaler(); Xtr=sc.fit_transform(Xtr); Xte=sc.transform(Xte)
base=LogisticRegression(max_iter=1000).fit(Xtr,ytr)
lda=LinearDiscriminantAnalysis(n_components=2); Ztr=lda.fit_transform(Xtr,ytr); Zte=lda.transform(Xte)
model=LogisticRegression(max_iter=1000).fit(Ztr,ytr)
print("Original accuracy:",round(accuracy_score(yte,base.predict(Xte)),4))
print("LDA accuracy:",round(accuracy_score(yte,model.predict(Zte)),4))
print("LDA explained variance:",lda.explained_variance_ratio_)
```

```python
plt.figure(figsize=(7,5))
for label in sorted(y.unique()):
    q=Ztr[ytr.to_numpy()==label]; plt.scatter(q[:,0],q[:,1],label=f"Class {label}")
plt.xlabel("Linear Discriminant 1"); plt.ylabel("Linear Discriminant 2")
plt.title("Iris Dataset after LDA"); plt.legend(); plt.grid(alpha=.25); plt.show()
print("Practical 2b Summary: LDA reduced four Iris features to two supervised discriminant components. It uses class labels to maximize class separation; accuracy is compared with the original-feature model.")
```
