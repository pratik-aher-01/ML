```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

p=list(Path(".").rglob("california_housing.csv"))
if not p: raise FileNotFoundError("Extract the practical datasets ZIP beside this notebook.")
df=pd.read_csv(p[0]); X=df.drop(columns="MedHouseVal"); y=df["MedHouseVal"]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
model=LinearRegression().fit(Xtr,ytr); pred=model.predict(Xte)
print("Features:",X.columns.tolist())
print("Intercept:",model.intercept_)
print("Coefficients:\n",pd.Series(model.coef_,index=X.columns))
print("MAE:",round(mean_absolute_error(yte,pred),4))
print("MSE:",round(mean_squared_error(yte,pred),4))
print("RMSE:",round(mean_squared_error(yte,pred)**.5,4))
print("R2:",round(r2_score(yte,pred),4))
```

```python
plt.figure(figsize=(6,6)); plt.scatter(yte,pred,alpha=.25)
lo,hi=yte.min(),yte.max(); plt.plot([lo,hi],[lo,hi],"r--")
plt.xlabel("Actual House Value"); plt.ylabel("Predicted House Value")
plt.title("Multiple Linear Regression"); plt.grid(alpha=.25); plt.show()
print("Practical 3 Summary: Multiple linear regression used all eight California Housing input features to predict median house value. MAE, MSE, RMSE and R-squared evaluated prediction quality.")
```
