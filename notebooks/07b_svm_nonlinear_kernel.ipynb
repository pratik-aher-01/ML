```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

p=list(Path(".").rglob("california_housing.csv"))
if not p: raise FileNotFoundError("Extract the practical datasets ZIP beside this notebook.")
df=pd.read_csv(p[0]).sample(n=4000,random_state=42); X=df.drop(columns="MedHouseVal"); y=df.MedHouseVal
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
model=make_pipeline(StandardScaler(),SVR(kernel="rbf",C=10,epsilon=.1)).fit(Xtr,ytr)
pred=model.predict(Xte)
print("MAE:",round(mean_absolute_error(yte,pred),4))
print("MSE:",round(mean_squared_error(yte,pred),4))
print("RMSE:",round(mean_squared_error(yte,pred)**.5,4))
print("R2:",round(r2_score(yte,pred),4))
```

```python
plt.figure(figsize=(6,6)); plt.scatter(yte,pred,alpha=.25)
plt.xlabel("Actual House Value"); plt.ylabel("Predicted House Value")
plt.title("RBF Nonlinear Kernel SVR"); plt.grid(alpha=.25); plt.show()
print("Practical 7b Summary: An RBF-kernel Support Vector Regression model predicted California house values. Standardization and regression metrics were used to assess nonlinear prediction performance.")
```
