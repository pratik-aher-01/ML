```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import adjusted_rand_score,silhouette_score

p=list(Path(".").rglob("make_moons.csv"))
if not p: raise FileNotFoundError("Extract the practical datasets ZIP beside this notebook.")
df=pd.read_csv(p[0]); X=df[["Feature_1","Feature_2"]]; y=df["True_Label"]
Xs=StandardScaler().fit_transform(X); display(df.head())
```

```python
labels=DBSCAN(eps=.25,min_samples=5).fit_predict(Xs); df["Cluster"]=labels
clusters=set(labels)-{-1}; noise=int((labels==-1).sum())
print("Clusters found:",len(clusters)); print("Noise points:",noise)
print("Adjusted Rand Index:",round(adjusted_rand_score(y,labels),4))
valid=labels!=-1
if len(set(labels[valid]))>1:
    print("Silhouette excluding noise:",round(silhouette_score(Xs[valid],labels[valid]),4))
else: print("Silhouette unavailable: fewer than two non-noise clusters.")
```

```python
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].scatter(X.iloc[:,0],X.iloc[:,1],c=y,cmap="viridis",s=18); ax[0].set_title("Original Labels")
ax[1].scatter(X.iloc[:,0],X.iloc[:,1],c=labels,cmap="viridis",s=18); ax[1].set_title("DBSCAN (-1 = Noise)")
for a in ax: a.set_xlabel("Feature 1"); a.set_ylabel("Feature 2")
plt.tight_layout(); plt.show()
print("Practical 9 Summary: DBSCAN identified density-based clusters and noise points in Make Moons data. The clusters were compared with supplied labels using adjusted Rand index.")
```
