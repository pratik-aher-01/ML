# Practical 09 — DBSCAN Clustering

Identify density-based groups and mark noise points with label -1.

## 1. Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
```

## 2. Create an open, reproducible synthetic clustering dataset

```python
X, _ = make_moons(n_samples=500, noise=0.07, random_state=42)
X_scaled = StandardScaler().fit_transform(X)
```

## 3. Apply DBSCAN

```python
model = DBSCAN(eps=0.25, min_samples=5)
labels = model.fit_predict(X_scaled)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = int(np.sum(labels == -1))
print("Number of clusters:", n_clusters)
print("Noise points:", n_noise)
```

## 4. Visualize clusters and noise

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap="viridis", s=25)
plt.xlabel("Feature 1 (scaled)")
plt.ylabel("Feature 2 (scaled)")
plt.title("DBSCAN clusters (-1 indicates noise)")
plt.show()
```

## 5. Evaluate silhouette when defined

```python
mask = labels != -1
clustered = labels[mask]
if len(set(clustered)) >= 2 and len(set(clustered)) < len(clustered):
    print("Silhouette score (excluding noise):", silhouette_score(X_scaled[mask], clustered))
else:
    print("Silhouette score is not defined for this clustering result.")
```
